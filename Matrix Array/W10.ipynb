### Stacking, Resize, Reshape

```python
x = np.array([[1,2,3],[5,6,7]])
y = np.array([2,3,4])
z = np.array([7,8,9])
xy = np.vstack((x,y,z))
print(xy)

## np.vstack menyusun array secara vertikal (baris ke bawah).
```

```python
a = np.array([[10],[11],[12],[13]])
A = np.hstack((xy,a))
A

## np.hstack menyusun array secara horizontal (ke samping).
```

```python
np.resize(A, (3,3))

## (3,3) -> potong/potong ulang data agar jadi 3×3.
```

```python
np.resize(A, (2,2))

 ## (2,2) -> potong jadi 2×2.
```

```python
np.resize(A, (4,5))

## (4,5) -> perulangan isi agar jadi 4×5 (jika data tidak cukup).
```

### Operasi Matrix

```python
mtx1 = np.arange(1,10).reshape((3,3))
mtx1

## Membuat array 1 sampai 9, lalu bentukkan menjadi matriks 3×3
```

```python
mtx2 = np.arange(21,30).reshape((3,3))
mtx2

## Membuat array 1 sampai 9, lalu bentukkan menjadi matriks 3×3
```

```python
jumlah  = mtx1 + mtx2
jumlah

## Penjumlahan elemen-elemen mtx1 dan mtx2 secara posisi
```

```python
kurang = mtx1 - mtx2;kurang

## Pengurangan elemen-elemen matriks
```

```python
skalar = mtx1 * mtx2;skalar

## Perkalian elemen per elemen (bukan perkalian matriks)
```

```python
perkalian = mtx1 @ mtx2;perkalian

## Perkalian matriks menggunakan operator @.
```

```python
perkalian2 = np.dot(mtx1, mtx2);perkalian2

## np.dot juga untuk perkalian matriks.
```

```python
np.transpose(perkalian)

## transpos dari hasil perkalian matriks: baris menjadi kolom dan sebaliknya.
```

```python
a = np.array([[10],[11],[12],[13]])
A = np.hstack((np.arange(1,13).reshape((4,3)),a));A

## Array dari 1 sampai 12 dibentuk menjadi 4×3, lalu digabung dengan kolom a sehingga menjadi 4×4.
```

```python
eigenvalue, eigenvector = np.linalg.eig(A)

## Menghitung nilai dan vektor eigen dari matriks A.
```

```python
eigenvalue
```

```python
eigenvector
```

```python
det = np.linalg.det(A);det

## Menghitung determinan dari matriks A.
```

```python
inv = np.linalg.inv(A);inv

## Menghitung invers dari matriks A.
```
